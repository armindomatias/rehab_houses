```python
# Test the LangGraph Agent with Idealista Scraper Tool

from src.agent_langgraph.agent import run_agent_test, create_agent
from langchain_core.messages import HumanMessage

# Example 1: Scrape a single listing
query1 = "Can you scrape the property data from https://www.idealista.pt/imovel/34458598/?"
print("Query 1:", query1)
print()
result1 = run_agent_test(query1)
print("Result:")
for msg in result1["messages"]:
    if hasattr(msg, 'content'):
        print(f"- {type(msg).__name__}: {msg.content[:200]}...")
print()
print("=" * 60)
print()
```

```python
# Example 2: Using the agent directly with more control
agent = create_agent()

# Create a conversation
state = {
    "messages": [
        HumanMessage(content="Hello! Can you help me scrape property data from Idealista?")
    ]
}

# Run the agent
result = agent.invoke(state)

# Print the conversation
for msg in result["messages"]:
    print(f"{type(msg).__name__}: {msg.content}")
    if hasattr(msg, 'tool_calls') and msg.tool_calls:
        print(f"  Tool calls: {msg.tool_calls}")
    print()
```
